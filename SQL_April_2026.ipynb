```python
# Code snippet 1
# Importing the dataset
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/Tarrant%20County%20Income%20Dataset%20SQL%20CSV.csv')

# Display the data
data
```

```python
# Code snippet 2
# Importing sqlite3 (to run SQL queries in Python)
import sqlite3
```

```python
# Code snippet 3
# Creating a sample (dummy) database as a connection
conn = sqlite3.connect('sample.db')
```

```python
# Code snippet 4
# Storing the pandas dataframe as a table in the sample database
data.to_sql('data_sql', conn)
```

```python
# Code snippet 5
# Selecting ALL the data from the data_sql table
sql_query = ''' SELECT * FROM data_sql '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 6
# Selecting ALL the customer data from the data_sql table whose age is more than 50
sql_query = ''' SELECT * FROM data_sql WHERE age > 50 '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 7
# Selecting ALL the customer data from the data_sql table whose age is <= 50
sql_query = ''' SELECT * FROM data_sql WHERE age <= 50 '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 8
# Selecting age, work class, education, marital status, income from the data_sql table
sql_query = ''' SELECT age, Workclass, Education, Marital_Status, Income FROM data_sql '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 9
# Selecting age, marital status from the data_sql table whose age < 30 and marital status is Never-married
sql_query = ''' SELECT age, Marital_Status FROM data_sql WHERE age < 30 AND marital_status = 'Never-married'  '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 10
# Selecting age, marital status from the data_sql table whose age > 40 and marital status is Married-civ-spouse
sql_query = ''' SELECT age, Marital_Status FROM data_sql WHERE age > 40 AND marital_status = 'Married-civ-spouse'  '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 11
# Selecting different/unique marital_statuses in the dataset
sql_query = ''' SELECT DISTINCT Marital_Status FROM data_sql '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 12
# Selecting different/unique marital_statuses and their RESPECTIVE counts in the dataset
sql_query = ''' SELECT Marital_Status, count(Marital_Status) FROM data_sql GROUP BY Marital_Status'''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 12.1
# Selecting different/unique marital_statuses and their RESPECTIVE counts in the dataset - order by the count descending
sql_query = ''' SELECT Marital_Status, count(Marital_Status) FROM data_sql GROUP BY Marital_Status ORDER BY count(Marital_Status) DESC '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 13
# Selecting different/unique education and their RESPECTIVE counts in the dataset - order by the count descending
sql_query = ''' SELECT education, count(education) FROM data_sql GROUP BY education ORDER BY count(education) DESC '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 14
# Selecting different/unique education, their RESPECTIVE counts and their average capital gains in the dataset - order by the count descending
sql_query = ''' SELECT education, count(education), avg(Capital_Gain) FROM data_sql GROUP BY education ORDER BY avg(Capital_Gain) DESC '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 15
# Selecting different/unique education, their RESPECTIVE counts and their average capital gains in the dataset WITH atleast 1000 instances /records- order by the count descending
sql_query = ''' SELECT education, count(education), avg(Capital_Gain) FROM data_sql GROUP BY education HAVING count(education) > 1000 ORDER BY avg(Capital_Gain) DESC '''
data_output = pd.read_sql(sql_query, conn)
data_output
```

```python
# Code snippet 16
# Selecting different/unique education, their RESPECTIVE counts and their average number of hours order by the count descending
sql_query = ''' SELECT education, count(education), avg(Hours_per_Week) FROM data_sql GROUP BY education ORDER BY avg(Hours_per_Week) DESC '''
data_output = pd.read_sql(sql_query, conn)
data_output
```
